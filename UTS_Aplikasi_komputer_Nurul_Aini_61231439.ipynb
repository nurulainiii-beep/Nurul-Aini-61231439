```python
import numpy as miayam

volume_kendaraan = miayam.array([120, 185, 150, 110, 135, 160, 145, 125, 175, 190])
print("sum:", miayam.sum(volume_kendaraan))
print("average:", miayam.average(volume_kendaraan))
print("max:", miayam.max(volume_kendaraan))
print("min:", miayam.min(volume_kendaraan))
```

```python
import pandas as pd

df = pd.read_csv("volume_kendaraan.csv", sep=";")
print("total volume kendaraan:", df["Volume Kendaraan"].sum())
print("rata-rata volume kendaraan:", df["Volume Kendaraan"].mean())
print("volume kendaraan tertinggi:", df["Volume Kendaraan"].max())
print("volume kendaraan terendah:", df["Volume Kendaraan"].min())
```

```python
import pandas as pd

df = pd.read_csv("volume_kendaraan.csv", sep=";", dtype={'Jam Pengamatan': str})
rata_rata = df['Volume Kendaraan'].mean()

jam_di_atas_rata_rata = df[df['Volume Kendaraan'] > rata_rata]

display(jam_di_atas_rata_rata[['Jam Pengamatan', 'Volume Kendaraan']])
```
